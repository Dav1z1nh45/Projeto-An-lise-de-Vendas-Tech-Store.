```python

#Lendo o arquivo CSV

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('vendas_techstore.csv')
print(df.head())


# Primeira inspeção dos dados

print("Shape:", df.shape)
print("\nTipos das colunas:")
print(df.dtypes)


# Convertendo a coluna de data

df['data'] = pd.to_datetime(df['data'])
df['mes'] = df['data'].dt.to_period('M').astype(str)

print(df.head())


# Melhor e pior vendedor

total_por_vendedor = df.groupby('vendedor')['valor'].sum().sort_values(ascending=False)
print(total_por_vendedor)

melhor_vendedor = total_por_vendedor.idxmax()
pior_vendedor = total_por_vendedor.idxmin()

print(f"\nMelhor vendedor: {melhor_vendedor} → R$ {total_por_vendedor[melhor_vendedor]:,.2f}")
print(f"Pior vendedor:   {pior_vendedor} → R$ {total_por_vendedor[pior_vendedor]:,.2f}")


# Região com maior e menor venda

total_por_regiao = df.groupby('regiao')['valor'].sum().sort_values(ascending=False)
print(total_por_regiao)

melhor_regiao = total_por_regiao.idxmax()
pior_regiao = total_por_regiao.idxmin()

print(f"\nRegião com maior venda: {melhor_regiao} → R$ {total_por_regiao[melhor_regiao]:,.2f}")
print(f"Região com menor venda: {pior_regiao} → R$ {total_por_regiao[pior_regiao]:,.2f}")


# Faturamento total e ticket médio

faturamento_total = df['valor'].sum()
ticket_medio = df['valor'].mean()

print(f"Faturamento total: R$ {faturamento_total:,.2f}")
print(f"Ticket médio:      R$ {ticket_medio:,.2f}")


# Ticket médio por vendedor e faturamento por mês

media_por_vendedor = df.groupby('vendedor')['valor'].mean().sort_values(ascending=False)
print("Ticket médio por vendedor:")
print(media_por_vendedor)

faturamento_por_mes = df.groupby('mes')['valor'].sum().sort_index()
print("\nFaturamento por mês:")
print(faturamento_por_mes)

print(f"\nMaior total: {melhor_vendedor}")
print(f"Maior ticket médio: {media_por_vendedor.idxmax()}")


# Gráficos

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(14, 10))

# Gráfico 01 - Total vendido por vendedor
cores_vendedor = [
    '#2E7D32' if v == melhor_vendedor else '#C62828' if v == pior_vendedor else '#90A4AE'
    for v in total_por_vendedor.index
]
ax1.bar(total_por_vendedor.index, total_por_vendedor.values, color=cores_vendedor)
ax1.set_title('Total vendido por vendedor', fontsize=13, fontweight='bold')
ax1.set_ylabel('Total (R$)')
ax1.tick_params(axis='x', rotation=0)

# Gráfico 02 - Total por região
cores_regiao = [
    '#2E7D32' if v == melhor_regiao else '#C62828' if v == pior_regiao else '#90A4AE'
    for v in total_por_regiao.index
]
ax2.bar(total_por_regiao.index, total_por_regiao.values, color=cores_regiao)
ax2.set_title('Total por região', fontsize=13, fontweight='bold')
ax2.set_ylabel('Total (R$)')
ax2.tick_params(axis='x', rotation=0)

# Gráfico 03 - Ticket médio por vendedor
ax3.bar(media_por_vendedor.index, media_por_vendedor.values, color='#1565C0')
ax3.set_title('Ticket médio por vendedor', fontsize=13, fontweight='bold')
ax3.set_ylabel('Ticket médio (R$)')
ax3.tick_params(axis='x', rotation=0)

# Gráfico 04 - Faturamento total por mês
ax4.bar(faturamento_por_mes.index, faturamento_por_mes.values, color='#EF6C00', width=0.5)
ax4.set_title('Faturamento total por mês', fontsize=13, fontweight='bold')
ax4.set_ylabel('Faturamento (R$)')
ax4.tick_params(axis='x', rotation=0)
```
